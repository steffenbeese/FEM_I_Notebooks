$\int f(x)dx = \frac{1}{10}x^5+3x^2+4x = [1/10*1^5+3*1^2+4*1]-[1/10*(-1)^5+3*(-1)^2+4*(-1)]$

Funktion 2: $\frac{3+x}{5+2(x-2)}$

```python
import numpy as np
import matplotlib.pyplot as plt

def fun_1(x):
    return 1/2*x**4+6*x+4
    
def fun_2(x):
    return (3+x)/(5+2*(x-2))

x = np.linspace(-1,1,100)
y1 = fun_1(x)
y2 = fun_2(x)

fi,ax = plt.subplots()
ax.plot(x,y2)

ax.set_title("Function 2")
ax.set_xlabel("x")
ax.set_ylabel("y")
```
